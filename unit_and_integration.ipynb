```python
import sys
import subprocess

subprocess.check_call([
    sys.executable,
    "-m",
    "pip",
    "install",
    "hypothesis",
    "pytest"
])

print("Installation completed")
import hypothesis
import pytest

print("Hypothesis version:", hypothesis.__version__)
print("Pytest installed successfully")
from hypothesis import given, strategies as st


def add(a: int, b: int) -> int:
    return a + b


def subtract(a: int, b: int) -> int:
    return a - b


def average(numbers: list[int]) -> float:
    if not numbers:
        raise ValueError("List cannot be empty")
    return sum(numbers) / len(numbers)


# Unit tests
def test_add():
    assert add(10, 20) == 30


def test_subtract():
    assert subtract(20, 10) == 10


def test_average():
    assert average([10, 20, 30]) == 20


# Hypothesis tests
@given(st.integers(), st.integers())
def test_add_property(a, b):
    assert add(a, b) == a + b


@given(st.integers(), st.integers())
def test_subtract_property(a, b):
    assert subtract(a, b) == a - b


# Integration test
def test_integration():
    total = add(10, 20)
    result = average([total, 20])
    assert result == 25


# Run tests
test_add()
test_subtract()
test_average()
test_add_property()
test_subtract_property()
test_integration()

print("All tests passed successfully!")
```
